```python
# listas para armazenar os melhores resultados
resultados_sphere = []

resultados_rastrigin = []

resultados_rosenbrock = []
```

```python
#sphere:
    #ciclos:
    #tc:
#rastrigin:
    #ciclos
    #tc:
#rosenbrock:
    #ciclos
    #tc:
CICLOS = 200
TAMANHO_COLONIA = 500
```

```python
from engine import executar_abc
from funcoes import sphere, rastrigin, rosenbrock
import matplotlib.pyplot as plt

DIMENSOES = 30

FATOR_ABANDONO = (
    TAMANHO_COLONIA // 2
) * DIMENSOES



def menu():

    print("--- simulador algoritmo abc ---")

    print("\nfunções disponíveis:")
    print("1 - Sphere")
    print("2 - Rastrigin")
    print("3 - Rosenbrock")

    escolha_f = input("escolha o número da função: ")

    if escolha_f == "1":

        funcao_ativa = sphere

        nome_funcao = "Sphere"

    elif escolha_f == "2":

        funcao_ativa = rastrigin

        nome_funcao = "Rastrigin"

    else:

        funcao_ativa = rosenbrock

        nome_funcao = "Rosenbrock"

    print(f"\nExecutando ABC para a função {nome_funcao}...")

    print("Aguarde.\n")

    final, historico = executar_abc(
        funcao_ativa,
        TAMANHO_COLONIA,
        DIMENSOES,
        CICLOS,
        FATOR_ABANDONO
    )

    # armazenamento do melhor resultado
    if escolha_f == "1":

        resultados_sphere.append(final)

    elif escolha_f == "2":

        resultados_rastrigin.append(final)

    elif escolha_f == "3":

        resultados_rosenbrock.append(final)

    # print("\n--- evolução da função objetivo ---")

    # for i, valor in enumerate(historico, start=1):

    #     print(f"Ciclo {i}: melhor valor = {valor}")

    print("\n" + "=" * 40)

    print("Resultado final:", final)
    print(f'quantidade de resultados sphere: {len(resultados_sphere)}')
    print(f'quantidade de resultados rastrigin: {len(resultados_rastrigin)}')
    print(f'quantidade de resultados rosenbrock: {len(resultados_rosenbrock)}')

    print("=" * 40)

    plt.plot(historico)

    plt.title(
        f"Gráfico de Convergência - {nome_funcao}"
    )

    plt.xlabel("Ciclos")

    plt.ylabel("Melhor Valor da Função")

    plt.grid(True)

    plt.show()
    


menu()
```

# Boxplot

```python
def gerar_boxplot(resultados_sphere,
                   resultados_rastrigin,
                   resultados_rosenbrock):

    dados = [
        resultados_sphere,
        resultados_rastrigin,
        resultados_rosenbrock
    ]

    nomes = [
        "Sphere",
        "Rastrigin",
        "Rosenbrock"
    ]

    plt.figure(figsize=(10, 6))

    plt.boxplot(dados, labels=nomes)

    plt.title("Comparação dos Resultados do ABC")

    plt.xlabel("Funções")

    plt.ylabel("Melhor Resultado")

    plt.grid(True)

    plt.show()
```

```python
gerar_boxplot(resultados_sphere,resultados_rastrigin,resultados_rosenbrock)
```
